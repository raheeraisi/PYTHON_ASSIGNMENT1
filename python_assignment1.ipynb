```python
string1="Hello "
string2=input("Enter your Name")
string3=string1+string2
print(string3)
string4=(",Welcome to Python Programming")
string5=string3+string4
print("original string:",string5)

# a.first character
print("First character:", string5[0])

#  b.Print the last character of the string. 
print("Last character:", string5[-1])

# c. Print the first 5 characters
print("First 5 characters:", string5[:5])

# d. Print the last 11 characters
print("Last 11 characters:", string5[-11:])

# e. Print the string in reverse
print("Reversed string:", string5[::-1])

# f. Print the word Python
print("Word Python:", string5[25:31])
```

# string methods

```python
strM = "Python beginner tutorial"

# a. Convert the sentence to uppercase
print("Uppercase:", strM.upper())

# b. Convert the sentence to lowercase
print("Lowercase:", strM.lower())

# c. Capitalize the sentence
print("Capitalized:", strM.capitalize())

# d. Count the occurrences of character 't'
print("Count of t:", strM.count('t'))

# e. Replace Python with Machine Learning
print("Replaced string:", strM.replace("Python", "Machine Learning"))
```

# Tuples (creation,modification,access)

```python
# Create two tuples
t1 = (10, 20, 30)
t2 = (40, 50, 60)

# a. Concatenate the two tuples
t_combine = t1 + t2
print("Combined tuple:", t_combine)

# b. Repeat the elements 3 times
print("Repeated tuple:", t_combine * 3)

# c. Access the 3rd element
print("3rd element:", t_combine[2])

# d. Access the first three elements
print("First three elements:", t_combine[:3])

# e. Access the last three elements
print("Last three elements:", t_combine[-3:])
```
